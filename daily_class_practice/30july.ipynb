### Built-in-function of list

#### 1.LEN ()

```python
a=[10,20,30,40,50]
print(len(a))
print(type(a))
```

#### 2max ()

```python
a1=[1,20,30,60,50]
print(max(a1))
print(type(a1))
```

### .6 all ()

```python
a3=[10,20,30,40,50]
print(all(a3))
print(type(a3))
```

### 7any ()

```python
a3=[10,20,30,40,50,1,0]
print(any(a3))
print(type(a3))
```

### 8. sorted ()

```python
a3=[10,20,30,40,50,5,6,9,56]
print(sorted(a3))
print(type(a3))
```

### 9.reverse ()

```python
a3=[10,20,30,40,50,5,6,9,56]
a2=list(reversed(a3))
print(a2)
print(type(a3))
```
