### TUPLE -  ()

### HOMOGENEOUS ELEMENT

```python
tuple1=(10,20,30,40,50)
print(tuple1)
```

#### HETROGENWOUS ELEMENT

```python
t=(10,20.3,25+6j,"hii",True)
print(t)
```

#### Duplicates element are allowed

```python
a=(10,20,50,20,10,46,96,9,100)
print(a)
print(type(a))
```

#### EMPTY TUPLE CREATION ()

```python
t=()
print(type(t))
```

### Creating tuple with only 1 element 9

```python
t1=(10,)
print(t1)
print(type(t1))
```

### Accessing Tuple ()

### 1. INDEXING ()

```python
t2=(10,20,30,40,50)
print(t2)
print(t2[2])
```

### 2. SLICING ()

```python
b=(10,20,30,40,50,50,60,90,155,15,24)
print(b[1:10:2])
print(type(b))
print(b[1])
print(b[1:6:1])
print(b[1:6:2])
```

## LOOPING ()

####  1. WHILE LOOP

```python
tuple1=(10,20,30)
index=0
while (index<len(tuple1)):
    print(tuple1[index])
    index=index+1
```

### 2. FOR LOOP ()

```python
tuple2=(10,20,30,40,50)
for i in tuple2:
    print(i)
```

### 3.ENNUMERATE ()

```python
tuple3=(10,20,30,40,50)
for index,value in enumerate(tuple3):
    print(index,value)
```

### BASIC TUPLE OPERATION 

### 1. IMMUTABILITY()

```python
c=(10,20,30,40,50)
print(c)
c[0]=100
print(c)
```

### 2. CONCATENATION ()

```python
a=(10,20,30,40,50)
b=(50,60,70,80,90)
print(b+a)
```
