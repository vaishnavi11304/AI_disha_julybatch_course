#### SET - Built-in function

### set are unorderd element - we cant use index and slicing here

#### 1. LEN {}

```python
set1={10,20,30,40,50}
print(len(set1))
```

#### 2.Max { }

```python
set1={10,20,30,40,50}
print(max(set1))
```

#### 3. Min { }

```python
set1={10,20,30,40,50}
print(min(set1))
```

#### 4. sum { }

```python
set1={10,20,30,40,50}
print(sum(set1))
```

#### 5. any ()

```python
set1={10,20,30,40,50,00,0,0}
print(any(set1))
```

#### 6. all { }

```python
set1={10,20,30,40,50}
print(all(set1))

a={1,2,3,4,5,0}
print(all(a))
```

#### 7.sorted { }

```python
set1={10,20,30,40,1,6,9,50}
print(sorted(set1))
```

## SET - METHOD

#### 1.ADD { }

```python
a={1,2,3,4,5,0}
a.add("vishu")
print(a)
```

#### 2.UPDATE { }

```python
set1={10,20,30,40,500}
set2={60,70,80}
set1.update(set2)
print(set1)
```

#### 3.REMOVE { }

```python
set1={10,20,30,40,50}
set1.remove(20)
print(set1)
```

#### 4.DISCARD { }

```python
set1={10,20,30,40,50}
set1.discard(20)
print(set1)
```

#### Remove vs Discard 

#### remove - if particular element is not available in the set , remove() method throws key-error  but , discard() method does not throw any type of error

#### 5. Clear ( )

```python
set1={10,20,30,40,50}
set1.clear()
print(set1)
```

#### 6. Copy { }

```python
set1={10,20,30,40,50}
set2=set1.copy()
print(set2)
print(id(set1))
print(id(set2))
```

### *DICTIONARY

```python
dict2={"Name":"Vishal","Age":22,"city":"Pune"}
print(dict2["Age"])
print(dict2.get("city"))
print(dict2.keys())
```

### using for loop - loops in dictionaries

```python
dict1={"name":"Vishal",
       "age":22,
       "city":"Pune"}
for k,v in dict1.items():
    print(k,v)
```

### dictionaries - print keys only { }

```python
dict1={"name":"Vishal",
       "age":22,
       "city":"Pune"}
for k,v in dict1.items():
    print(k)
```

### Print items only { }

```python
dict1={"name":"Vishal",
       "age":22,
       "city":"Pune"}
for k,v in dict1.items():
    print(v)
```

### Enumerate { }

```python
dict4={"name":"Vishal",
       "age":22,
       "city":"Pune"}
for i, (k, v) in enumerate(dict4.items(),start=1):
    print(i, k, v)
```
