## aliasing and cloning

```python
list1=[1,2,3,4,5,6,7,556]
print(id(list1))
list2=list1
print(id(list2))
list2[1]=21
print(list1)
```

### clonning

```python
list1=[10,20,30,40,50]
list2=list1.copy()
list1[0]=11
print(list1)
print(list2)
```
